```python
import numpy as np

A = np.array((
    [1,2],
    [3,4]
))

eigenvalues, eigenvectors = np.linalg.eig(A)

print(eigenvalues)
```

```python
print(eigenvectors)
```

```python
U, S, V = np.linalg.svd(A)

print("U matrix:\n", U)
print("Sigma values:\n", S)
print("V matrix:\n", V)
```

```python
print(U @ U.T)
```
