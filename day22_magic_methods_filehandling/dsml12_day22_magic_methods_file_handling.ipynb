```python
class BankCustomer:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def display(self):
    return f"{self.name} is {self.age} years old"

  # called by print()
  def __str__(self):
    return f"{self.name} is {self.age} years old"

  # called by repr
  def __repr__(self):
    return f"{self.name} is {self.age} years old"

  # called by len
  def __len__(self):
    return self.age

  # makes object callable like a function
  def __call__(self, gender = "male"):
    return f"{self.name} is {self.age} years old and {gender}"
```

```python
customer = BankCustomer("John", 30)
```

```python
# __str__
print(customer)
```

```python
customer.display()
```

```python
# __repr__
print(repr(customer))
```

```python
# __len__
print(len(customer))
```

```python
# __call__
print(customer())
```

```python
print(customer("other"))
```
