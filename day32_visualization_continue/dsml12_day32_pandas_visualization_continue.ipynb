```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
x = np.random.standard_t(500, size=1000)
```

```python
x[:15]
```

image.png

```python
_mean = np.mean(x)
```

```python
print(f"Mean of x: {_mean}")
```

```python
plt.hist(x, bins=20, color="blue")
```

```python
plt.axvline(x=_mean, color="red")
```

```python
plt.hist(x, bins=20, color="blue")
plt.axvline(x=_mean, color="red")
```

Screenshot 2026-06-29 at 17.05.03.png

```python
_median = np.median(x)
print(f"Median of x: {_median}")
```

```python
plt.hist(x, bins=20, color='blue')
plt.axvline(x=_median, color='yellow')
```

```python
plt.hist(x, bins=20, color='blue')
plt.axvline(x=_mean, color='red')
plt.axvline(x=_median, color='yellow')
```

Screenshot 2026-06-29 at 17.05.58.png

```python
_mode = stats.mode(x).mode
print("Mode:",_mode)
```

```python
plt.hist(x, color="blue")
plt.axvline(_mode, color="red")
plt.show()
```
