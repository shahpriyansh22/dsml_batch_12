```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print(self.name, self.age)

class Student(Person):
    def __init__(self, name, age, student_id):
        Person.__init__(self, name, age)
        self.student_id = student_id

    def study(self):
        print("Student ID:", self.student_id)

class Teacher(Person):
    def __init__(self, name, age, subject):
        Person.__init__(self, name, age)
        self.subject = subject

    def teach(self):
        print("Subject:", self.subject)

class TeachingAssistant(Student, Teacher):
    def __init__(self, name, age, student_id, subject, department, salary):
        Student.__init__(self, name, age, student_id)
        Teacher.__init__(self, name, age, subject)
        self.department = department
        self.salary = salary

    def assist(self):
        print("Department:", self.department)
        print("Salary:", self.salary)
```

```python
ta = TeachingAssistant(
    "Bibek",
    22,
    "ST101",
    "Python",
    "Computer Science",
    25000
)

ta.show_person()
ta.study()
ta.teach()
ta.assist()
```

```python
class Employee:
  def __init__(self, nm, sal):
    self.name = nm
    self.salary = sal

  def getName(self): # camel case
    return self.name

  def getSalary(self):
    return self.salary
```

```python
class Salesoffice(Employee):
  def __init__(self, nm, sal, inc):
    super().__init__(nm, sal)
    self.incnt = inc

  def getSalary(self):
    return self.salary + self.incnt
```

```python
e1 = Employee("Priyansh", 900000)
print("Total salary of {} is Rs {}".format(e1.getName(), e1.getSalary()))
```

```python
s1 = Salesoffice("Ram", 8000000, 100000)
print("Total salary of {} is Rs {}".format(s1.getName(), s1.getSalary()))
```

```python
class Example:
  def __init__(self):
    self.public_var = 1 # Public variable
    self.__private_var = 2 # Private variable
    self._protected_var = 3 # Protected variable

  def public_method(self):
    print("Public method")
    print("Private variable", self.__private_var)

  def __private_method(self):
    print("Private method")

  def _protected_method(self):
    print("Protected method")
```

```python
# creating object

obj = Example()
```

```python
# accessing public members

print(obj.public_var)
obj.public_method()
```

```python
# accessing protected members (possible not recommended)
print(obj._protected_method)
obj._protected_method()
```

```python
# accessing private members directly
# print(obj.__private_var)
# obj.__private_method
```

```python
# Accessing private members using name mangling

print(obj._Example__private_var)
obj._Example__private_method()
```

```python
# Class Task

class BankAccount:
  def __init__(self, name, initial_balance):
    self.name = name
    self.__balance = initial_balance # private variable

  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      return True
    return False

  def withdraw(self, amount):
    if 0 < amount <= self.__balance:
      self.__balance -= amount
      return True
    return False

  def check_balance(self):
    return self.__balance

class SavingsAccount(BankAccount):
  def __init__(self, name, initial_balance, interest_rate):
    super().__init__(name, initial_balance)
    self.interest_rate = interest_rate

  def calculate_interest(self):
    interest = self.check_balance() * self.interest_rate/100
    self.deposit(interest)
    return interest

account = SavingsAccount("Priyansh", 1000, 5)

print("Initial balance:", account.check_balance())
account.deposit(500)
print("After deposit:", account.check_balance())
account.withdraw(300)
print("After withdrawal:", account.check_balance())
interest = account.calculate_interest()
print("Interest added:", interest)
print("Final balance:", account.check_balance())
```

```python
#

class BankAccount:
  def __init__(self, name, initial_balance):
    self.name = name
    self._account_type = "Bank"
    self.__balance = initial_balance

  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
    else:
      print("Invalid deposit amount")

  def withdraw(self, amount):
    if amount <= 0:
      print("Invalid withdrawal amount")
    elif amount > self.__balance:
      print("Insufficient balance! Cannot withdraw")
    else:
      self.__balance -= amount
      print(f"withdrawn: ${amount}")
  def check_balance(self):
    print(f'CurrentBalance: ${self.__balance}')

  def get_initial_balance(self):
    return self.__balance

class SavingAccount(BankAccount):
  def __init__(self, name, initial_balance, interest_rate):
    super().__init__(name, initial_balance)
    self._account_type = "Saving"
    self.__interest_rate = interest_rate

  # def add_interest(self):
  #   interest = self.get_initial_balance()
  #   print(interest)
  #   self.__balance += interest
  #   print(f"Interest added: ${interest}")

  def add_interest(self):
    interest = self.get_initial_balance() * (self.__interest_rate / 100) # Calculate actual interest
    print(f"Calculated interest: ${interest}")
    self.deposit(interest) # Use deposit method to update balance
    print(f"Interest added: ${interest}")

  def account_info(self):
    print("Account Holder:", self.name)
    print("AccountType:",self._account_type)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("Woof!")

class Cat(Animal):
  def make_sound(self):
    print("Meow!")
```

```python
dog = Dog()
dog.make_sound()
```

```python
cat = Cat()
cat.make_sound()
```
