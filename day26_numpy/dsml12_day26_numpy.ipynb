```python
# List vs Numpy array
```

```python
# List

import time
start = time.time()
py_list = [i*2 for i in range(100000000)]
print("\n List operation time:", time.time()-start)
```

```python
# Numpy

start = time.time()
np_array = np.arange(100000000)*2
print("\n Numpy operation time:", time.time()-start)
```

```python
# List

py_list = [1,2,3]
print("Python list multiplication", py_list*3)
```

```python
ny_array = np.array([1,2,3]) # Element wise multiplication
print("Numpy array multiplication", ny_array*3) # 3 is called scalar, we can perform the vector multiplication
```
