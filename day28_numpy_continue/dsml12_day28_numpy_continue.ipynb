```python
import numpy as np
```

```python
matrix_1 = np.array([
    [1,2],
    [3,4]
])
matrix_1
```

```python
matrix_2 = np.array([
    [4,6],
    [7,8]
])
matrix_2
```

```python
result = matrix_1 + matrix_2 # element wise addition
result
```

```python
matrix_sub = matrix_1 - matrix_2 # element wise subtraction
matrix_sub
```

```python
matrix_mux = matrix_1 * matrix_2 # wrong concept of matrix multiplication
matrix_mux
```

```python
matrix_mx = matrix_1 @ matrix_2 # correct method of matrix multiplication
matrix_mx
```

```python
matrix_mx = np.matmul(matrix_1, matrix_2)
matrix_mx
```

```python
result_div = matrix_1 / matrix_2 # not recommended
result_div
```

```python
# Element wise operations with scalars

print(matrix_1)
print("//////////")
scalar = 2
result_add = matrix_1 + scalar
print(result_add)
```

```python
print(matrix_1)
print("//////////")
scalar = 2
result_mul = matrix_1 * scalar
print(result_mul)
```

```python
# Broadcasting

matrix_3 = np.array([
    [1,2],
    [3,4]
])
array_1 = np.array([5,6])
result = matrix_3 + array_1
result
```

```python
# Computing the determinant

A = np.array([
    [1,2],
    [3,4]
])

det = np.linalg.det(A)
print(det)
```

```python
# Computing the inverse

A_inv = np.linalg.inv(A)
print(A_inv)
```

```python
print(A @ A_inv)
```

```python
a = np.array([
    [2,1],
    [1,3]
])

b = np.array([5,6])
X = np.linalg.solve(a,b)
X
```

```python
# class task

# x+y+z=6
# 2x+5y+z=-4
# 2x+3y+8z=5

a = np.array([
    [1,1,1],
    [2,5,1],
    [2,3,8]
])

b = np.array([6,-4,5])
X = np.linalg.solve(a,b)
X
```

```python
#

X = np.array([
    [1,1],
    [2,1],
    [3,1]
])

y = np.array([2,3,5])

beta = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
beta
```

```python
print("Slope (m):", beta[0])
```

```python
print("Intercept (c):", beta[1])
```
