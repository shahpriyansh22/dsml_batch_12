```python
import numpy as np

fare_bins = [0,50,100,300,512]
fare_labels = ["Low Fare", "Medium Fare", "High Fare", "Very High Fare"]

titanic_df["FareGroup"] = pd.cut(
    titanic_df["Fare"],
    bins=fare_bins,
    labels=fare_labels,
    include_lowest=True
)
```

```python
# Bar chart
# Gender count

import matplotlib.pyplot as plt
```

```python
titanic_df["Sex"].value_counts().plot.bar()
plt.title("Gender Distribution")
plt.show()
```

```python
titanic_df['Survived'].value_counts().plot(
    kind="bar",
    title="Count of survived Passengers",
    figsize=(5,5),
    legend=True,
    yticks=list(range(0,600,50)),
    xlabel="Survived",
    ylabel="Count",
    grid=True,
    color="red"
)
```

image.png

```python
pd.crosstab(titanic_df.Survived, titanic_df.Sex).plot(
    kind="bar",
    title = "count of survived passengers",
    figsize= (5,5),
    legend=True,
    xlabel="Survived (0=No, 1=yes)",
    ylabel="No. of Passengers",
    grid=True,
    colormap="Dark2"
)
```
