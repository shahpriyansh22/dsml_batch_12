```python
# Factorial number

# 5! = 5*4*3*2*1 = 120
# 7! = 7*6*5!

# 1. n * (n-1) (recursive case)
# 2. 1 or 0 (base case)
```

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * factorial(n-1)
```

```python
factorial(5)
```

```python
# Decorator

def myWrapper(func):
  def myInnerFunc():
    print("Inner wrapper.")
    func()
  return myInnerFunc
```

```python
@myWrapper
def myFunc():
  print("Hello, world!")
```

```python
myFunc()
```

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"Calling {func.__name__}")
    return func(*args, **kwargs)
  return wrapper
```

```python
@logger
def greet(name,age):
  print(f"Hello {name}, Age: {age}")
```

```python
greet("Priyansh",age = 22)
```
