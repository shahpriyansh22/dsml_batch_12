```python
# Student Grade Management System

class Student:
    # Class attribute
    school = "SkillShikshya School"

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance method
    def check_grade(self):
        if self.marks >= 80:
            print(f"{self.name}'s Grade: A")
        elif self.marks >= 60:
            print(f"{self.name}'s Grade: B")
        elif self.marks >= 40:
            print(f"{self.name}'s Grade: C")
        else:
            print(f"{self.name}'s Grade: Fail")

    # Class method
    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)

    # Static method
    @staticmethod
    def welcome():
        print("Welcome to the Student Grade Management System!")


# Create an object
student1 = Student("Priyansh", 75)

# Call static method
Student.welcome()

# Call class method
Student.display_school()

# Display student's grade
student1.check_grade()
```

```python
# Bank Account Management System

class BankAccount:
    # Class attribute
    bank_name = "SkillShikshya Bank"

    # Constructor
    def __init__(self, holder_name, balance):
        self.holder_name = holder_name
        self.balance = balance

    # Instance method to deposit money
    def deposit(self):
        amount = float(input("Enter deposit amount: "))
        self.balance += amount

    # Instance method to withdraw money
    def withdraw(self):
        amount = float(input("Enter withdrawal amount: "))
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient Balance!")

    # Instance method to display balance
    def show_balance(self):
        print("Current Balance:", self.balance)

    # Class method
    @classmethod
    def display_bank_name(cls):
        print("Bank Name:", cls.bank_name)

    # Static method
    @staticmethod
    def welcome():
        print("Welcome to SkillShikshya Bank")


# ---------------- Main Program ----------------

# Call static method
BankAccount.welcome()

# Input account details
name = input("Enter account holder name: ")
balance = float(input("Enter initial balance: "))

# Create object
account = BankAccount(name, balance)

# Display bank name
BankAccount.display_bank_name()

# Deposit money
account.deposit()
account.show_balance()

# Withdraw money
account.withdraw()
account.show_balance()
```
