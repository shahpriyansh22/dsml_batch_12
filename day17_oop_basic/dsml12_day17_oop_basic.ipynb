```python
class Character:
  def __init__(self, health, damage, speed): # Instance method (Constructor)
    self.health = health # Instance attributes
    self.damage = damage
    self.speed = speed

  def double_speed(self): # Instance method
    self.speed *= 2
    return self.speed
```

```python
obj1 = Character(100,50,40)
obj1.speed
```

```python
obj1.double_speed()
```

```python
obj2 = Character(30,40,50)
obj2.double_speed()
```

```python
class Calculator:

  # A utility function that logically belong to the class but doesnot need self or cls

  @staticmethod # Static method (Decorator)
  def add(a,b):
    return a+b
```

```python
Calculator.add(3,4)
```

```python
class EmailValidator:

  @staticmethod
  def is_valid(email):
    return "@" in email
```

```python
EmailValidator.is_valid("test@gmail.com")
```

```python
# Class Task (Slide page no. 12)

class Calculator:

  company = "Priyansh Python Calculator"

  def __init__(self):
    self.num1 = float(input("Enter first number: "))
    self.num2 = float(input("Enter second number: "))
    self.operator = input("Enter operator (+, -, *, /): ")

  def calculate(self):
    if self.operator == "+":
      print("Result:", self.num1 + self.num2)

    elif self.operator == "-":
      print("Result:", self.num1 - self.num2)

    elif self.operator == "*":
      print("Result:", self.num1 * self.num2)

    elif self.operator == "/":

      if self.num2 != 0:
        print("Result:", self.num1 / self.num2)
      else:
        print("Cannot divide by zero")

    else:
      print("Invalid operator")

  @classmethod
  def show_company(cls):
    print("Company Name:", cls.company)

  @staticmethod
  def welcome():
    print("Welcome to Priyansh Python Calculator Program!")

calc = Calculator()

calc.welcome()

calc.show_company()

calc.calculate()
```
