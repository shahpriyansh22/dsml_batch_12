```python
# Class Task (Slide page no. )

class Polygon:
  def __init__(self, name, sides):
    self.name = name
    self.sides = sides
  def display(self):
    print(f"This is {self.name} and it has {self.sides} sides.")
```

```python
triangle = Polygon("Triangle", 3)
triangle.display()
```

```python
square = Polygon("Square", 4)
square.display()
```

```python
# Traditional Method Overloading

class Calculator:

  def add(self,a,b):
    return a+b

  def add(self,a,b,c):
    return a+b+c
```

```python
obj = Calculator()
obj.add(1,2)
```

```python
# Using Default Argument

class Calculator:

  def add(self,a,b=0,c=0):
    return a+b+c
```

```python
obj1 = Calculator()
obj1.add(10,20,10)
```

```python
obj1.add(10,20)
```

```python
# Multilevel Inheritance

class Vehicle:
    def move(self):
        return "Moving on wheels..."

class Car(Vehicle):
    def honk(self):
        return "Beep beep!"

class ElectricCar(Car):
    def charge(self):
        return "Charging the battery..."

my_car = ElectricCar()

print(my_car.move())
print(my_car.honk())
print(my_car.charge())
```

```python
# Heirarchical Inheritance

class Vehicle:
    def info(self):
        return "This is a vehicle."

class Car(Vehicle):
    def drive(self):
        return "Driving on four wheels."

class Motorcycle(Vehicle):
    def ride(self):
        return "Riding on two wheels."

my_car = Car()
my_bike = Motorcycle()

print(my_car.info())
print(my_car.drive())

print(my_bike.info())
print(my_bike.ride())
```

```python
#

class OTTPsubscription:
  def __init__(self, subscription_id, plan, total_payment):
    self.id = subscription_id
    self.plan_package = plan
    self.total_payment = total_payment

  def subscription(self):
    print(f"Subscriber with {self.id} id subscribed to the {self.plan_package}")

  def unsubscribe(self):
    print(f"Subscriber with {self.id} unsubscribed to the {self.plan_package}")
```

```python
class PremiumSubscription(OTTPsubscription):
  def __init__(self, subscription_id, plan, total_payment, screens):
    super().__init__(subscription_id, plan, total_payment)
    self.max_screen = screens

  def set_max_screen(self):
    print(f"Maximum screen set to {self.max_screen} in the premium plan")
```

```python
obj = PremiumSubscription(1222, "Pro", 800000, 7)
```

```python
obj.set_max_screen()
```

```python
obj.subscription()
```

```python
#

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):
    def __init__(self, name, age, student_id):
        Person.__init__(self, name, age)
        self.student_id = student_id

    def show_student(self):
        print("Student ID:", self.student_id)


class Teacher(Person):
    def __init__(self, name, age, subject):
        Person.__init__(self, name, age)
        self.subject = subject

    def show_teacher(self):
        print("Subject:", self.subject)


class TeachingAssistant(Student, Teacher):
    def __init__(self, name, age, student_id, subject, department, salary):
        Student.__init__(self, name, age, student_id)
        Teacher.__init__(self, name, age, subject)
        self.department = department
        self.salary = salary

    def show_ta(self):
        print("Department:", self.department)
        print("Salary:", self.salary)


ta = TeachingAssistant("Priyansh", 24, "8656", "DSML", "IT", 25000)

ta.show_person()
ta.show_student()
ta.show_teacher()
ta.show_ta()
```
